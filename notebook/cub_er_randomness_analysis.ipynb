# CUB ER Graph Randomness Analysis

Compare original CREAM CUB graph statistics to an ER/fixed-edge random graph ensemble.

```python
from pathlib import Path
import pandas as pd

root = Path('../experiments/CUB/graph_randomness_er')
z = pd.read_csv(root / 'z_scores.csv')
ensemble = pd.read_csv(root / 'er_ensemble_metrics.csv')
original = pd.read_csv(root / 'original_metrics.csv').iloc[0]
z.sort_values('z_score', key=lambda s: s.abs(), ascending=False)
```

```python
import matplotlib.pyplot as plt

ax = z.assign(abs_z=z['z_score'].abs()).sort_values('abs_z').plot.barh(
    x='metric', y='z_score', legend=False, figsize=(8, 4)
)
ax.axvline(2, color='tab:red', linestyle='--', alpha=0.6)
ax.axvline(-2, color='tab:red', linestyle='--', alpha=0.6)
ax.axvline(3, color='tab:purple', linestyle=':', alpha=0.6)
ax.axvline(-3, color='tab:purple', linestyle=':', alpha=0.6)
ax.set_xlabel('z-score of original graph vs ER ensemble')
ax.set_ylabel('metric')
ax.set_title('How unusual is the original CUB graph?')
plt.tight_layout()
```

```python
metrics = z['metric'].tolist()
fig, axes = plt.subplots(2, 3, figsize=(12, 7))
axes = axes.ravel()
for ax, metric in zip(axes, metrics):
    ax.hist(ensemble[metric], bins=20, alpha=0.75)
    ax.axvline(original[metric], color='tab:red', linewidth=2, label='original')
    ax.set_title(metric)
    ax.legend()
plt.tight_layout()
```
